# 1.0 Modelo

## 1.1 KMean

```python
K = 5
kmeans = KMeans( init='random', n_clusters=K, n_init=10, max_iter=100, random_state=42)

kmeans.fit(df)

labels = kmeans.labels_
```

### 1.1.1 Métricas

#### 1.1.1.1 WSS

```python
kmeans.inertia_
```

#### 1.1.1.2 SS

```python
m.silhouette_score(df, labels, metric='euclidean')
```
