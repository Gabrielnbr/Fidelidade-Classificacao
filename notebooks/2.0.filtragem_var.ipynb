# 1.0 Tratamento

## 1.1 Numerico

### 1.1.1 unit_price

Os valores zerados são assumidos como brindes.

```python
df = df.loc[df['unit_price'] >= 0.04,:]
```

## 1.2 Categórico

### 1.2.1 stock_code

Retirar valores indevidos

```python
lista_stock_code = ['POST', 'D', 'C2', 'DOT', 'M', 'BANK CHARGES', 'S', 'AMAZONFEE', 'DCGS0076', 'DCGS0003', 'gift_0001_40', 'DCGS0070', 'm', 'gift_0001_50', 'gift_0001_30', 'gift_0001_20', 'DCGS0055', 'DCGS0072', 'DCGS0074', 'DCGS0069', 'DCGS0057', 'DCGSSBOY', 'DCGSSGIRL', 'gift_0001_10', 'PADS', 'DCGS0004', 'DCGS0073', 'DCGS0071', 'DCGS0068', 'DCGS0067', 'DCGS0066P', 'B', 'CRUK']
```

```python
df = df.loc[~(df['stock_code'].isin(lista_stock_code))]
```

### 1.2.2 description

não irei trabalhar com eles agora

```python
df = df.drop(columns=['description'])
```

### 1.2.3 Country

Retirar paises não identificados e Comunidade europeia, por não ser país

```python
df = df[~(df['country'].isin(['European Community','Unspecified']))]
```

### 1.3 Separação em 2 DF Compra e Devolução

```python
df_returns = df.loc[df['quantity'] < 0,:]
df_compras = df.loc[df['quantity'] > 0,:]
```
